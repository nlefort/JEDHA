```python
import openai
import os
from dotenv import load_dotenv

load_dotenv()
openai.api_key = os.getenv("OPENAI_API_KEY")  # <-- ici OPENAI_API_KEY

print("API Key:", openai.api_key)  # si None => problème de chargement du .env


def generate_quiz(theme: str):
    system_prompt = (
        "Tu es un professeur d'histoire-géographie pour des élèves de CM1. "
        "Ton but est de générer des questions simples, claires, compréhensibles pour un enfant de 9-10 ans."
    )

    user_prompt = (
        f"Génère 3 questions à choix multiples sur le thème suivant : \"{theme}\".\n"
        "Pour chaque question, donne 3 propositions (a, b, c) et indique la bonne réponse."
    )

    response = openai.ChatCompletion.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
        ],
        temperature=0.7, #
        max_tokens=800,
    )

    return response.choices[0].message.content.strip()

# Exemple
if __name__ == "__main__":
    theme = input("Thème du quiz (ex: Qui sont les Gaulois ?) : ")
    quiz = generate_quiz(theme)
    print("\n--- Quiz généré ---\n")
    print(quiz)
```
