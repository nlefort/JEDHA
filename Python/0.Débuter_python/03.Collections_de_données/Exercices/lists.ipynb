```python
"""pour créer une liste il nous faut nos 2 crochets []"""
names=["Antoine", "Michelle", "Doug", "William", "Julia"]
print(names)
```

```python
names[2] # Si je veux atteindre l'un des noms, je cherche par l'index
```

```python
# Ajouter un nom à ma liste un nom
names.append("Lani")
print(names)
```

```python
#insert() pour ajouter des items ailleurs qu'en fin de liste
list=[1, 2, 5, 89, 3]
list.insert(3,"la vie") #2 arguments : l'index dans lequel je souhaite insérer mon item + item
list
```

```python
#pop() avec la fonction pop(), je supprime un élément
print(names)
doug=names.pop(3) #par défaut pop() prend le dernier élément, mais je peux lui indiquer l'index()
print(names)
print(doug) #pop() garde en mémoire ce qui a été supprimé

#pop() permet aussi de déplacer un item dans une liste
list=[3, 1, "hello"]
list.insert(1, list.pop()) #je met 1 à la fin de ma liste
print(list)
```

```python
#je peux remplacer par un autre nom (William par Will)
names[3]="Will"
print(names)
```

```python
#remove() je supprime un élément de ma liste
list=[1, 2, 5, 89, 3]
list.remove(2) #peut permettre d'enlever les redondances
print(list)
```
