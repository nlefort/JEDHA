```python
#autre moyen de faire qui n'est pas plus simple
comp=[]
for i in range(10):
    comp.append(i)
print(comp)
```

```python
[i+10 for i in range(10)]
```

```python
greetings=["Hello", "World", "From", "Jedha"]
[(i+10, hello) for i, hello in enumerate(greetings)]
```

```python
greetings=["Hello", "World", "From", "Jedha"]
comp=[]
for i, hello in enumerate(greetings):
    comp.append((i+10,hello))
print(comp)
```

```python
#list comprehension

comp=[i for i in range(10)]
print(comp)
```

```python
["even" if number %2==0 else "odd" for number in comp]
```

```python
#on peut imbriquer autant de if que l'on veut 
["even" if number %2==0 else "odd" if number %2==1 else "IMPOSSIBLE" for number in comp]
```

```python
# attention à l'écriture
# j'ai filtré tout les odd numbers
["even" for number in comp if number %2==0]
```
