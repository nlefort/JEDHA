```python
#le dictionnaire se caractérise par des clés et des valeurs associées à ces clés. chaque clé sera séparée par une virgule
#friend = nom de ma variable
#name = key / value of key name : Antoine. Ne pas mettre '' si value = int
friend={"Name":"Antoine", "Amount":112}
print(friend)
```

```python
#pour accéder à ma clé name
friend["Name"]
```

```python
#je peux aussi modifier la valeur de ma clé
friend["Amount"]=49
print(friend)
```

```python
#ajouter une nouvelle clé en ajoutant une nouvelle chaine dans mes {}
friend["last_name"]="Lefort"
print(friend)
```

```python
#je peux bouger les éléments comme pour une liste
last_name=friend.pop("last_name")
print(friend)
print(last_name)
```

```python
#how much did they spent ? et le stocker dans le dictionnaire
friends=input("what are the names of your friends? (separate their names by a coma")
amounts=input("how much each of your friends spent (keep the same order as above and separate by a coma)")
```

```python
list_of_friends=friends.split(",")
list_of_amounts=amounts.split(",")
```

```python
friends=[
    {"Name":list_of_friends[0], "Amount":list_of_amounts[0]},
    {"Name":list_of_friends[1], "Amount":list_of_amounts[1]}
]
print(friends)
```
