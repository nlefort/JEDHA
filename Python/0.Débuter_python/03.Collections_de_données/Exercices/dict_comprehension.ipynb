```python
#je crée un dictionnaire
stocks={"APPLE":274, "GOOGLE":1250, "MICROSOFT":2369}
print(stocks)
```

```python
#j'associe chaque clé et chaque valeur
for key, value in stocks.items():
    print(key,value)
```

```python
#je sépare mes clés/valeurs
for key in stocks.keys():
    print(key)
for value in stocks.values():
    print(value)
```

```python
#je fais un calcul dans mon dictionnaire
dollar_to_euro=0.92
{key:value*dollar_to_euro for key, value in stocks.items()}
```

```python
#j'exclue une valeur de mon dictionnaire
{key:value for key, value in stocks.items() if value > 1000}
```

```python
je qualifie chaque clé de mon dictionnaire
{key:("low price" if value < 1000 else "high price") for key, value in stocks.items()}
```
