```python
names=["Julia", "Antoine", "Adele", "Alex","Thomas" ] #liste de prénoms

for i, name in enumerate(names): #pour chaque répétition, prénom dans ma liste, je les énumère
    search_name="Adele" #je cherche Adèle
    print(name) #j'écris Adèle
    if name == search_name : #si le prénom == le prénom cherché
        print("{} is at index {}".format(search_name,i)) #j'écris le prénom cherché est de la place de la répéTition n°...
        break #j'arrète d'énumérer dès qu'Adèle est trouvée
```
