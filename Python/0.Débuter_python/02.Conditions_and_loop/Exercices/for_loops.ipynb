```python
items=["Hello", "world", "from", "jedha"] #items = nom de ma liste // j'écrit une liste [] // mes variables sont des chaînes de caractères ""
print(items)
```

```python
print(items[0]) #j'imprime le premier mot de ma liste 
print(items[1]) # et ainsi de suite 
print(items[2])
print(items[3])
# mais c'est pas très "dry" parce que je répète 3 fois la même chose finalement
```

## For loops ##

```python
for item in items: #la convention veut que j'utilise le singulier du nom de ma variable que je souhaite itérer
    print(item)
```

```python
#autre moyen pour atteindre le même résultat
#ma variable qui s'appelle "longueur". je veux aller chercher tous les éléments len() de ma variable items
#pour chaque valeur dans ma liste items, j'écris ce qu'il y a dedans
#range prend le nombre comme un argument et créer une liste depuis 0 jusqu'à ce qu'on dise stop
length = len(items)
for i in range(length):
    print(items[i])
```

```python
for i in range (10): #range -> valeurs numériqes. pour chaque valeur présente dans ma liste (0->10) j'écris ce qu'il y a dedans et je m'arrête avant 10
    print(i)
```

```python
for item in enumerate(items):# quand on fait des listes on travaille plutôt avec enumerate.
        print(item)
```

```python
#Toujours mettre ':' après loops
for i, item in enumerate(items):
    print("Item {} has index {}".format(item,i))
```
