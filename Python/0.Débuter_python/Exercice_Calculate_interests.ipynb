# How much do you get for your savings ?

It is said that if Christopher Columbus had set aside €1 that he had left at 2% interest per year, today his heirs would no longer need to work to live their lives. Let's try to calculate the same thing for our savings. 

1. So let's create a function that will allow us to know how much money a user will have in total after a certain number of years.

2. The user should be able to call the function which will then ask for :

    1. The total amount he wishes to place
    2. The number of years he wants to invest his money for
    3. The rate of interest to which he is entitled
    
3. Pay attention to the errors that the user may enter.

    4. If he enters a string of characters rather than a number, an error should be displayed.
    5. If the user enters negative numbers, an error should also be output.
    6. Finally, it is very likely that the user will write the interest rate as a percentage (e.g. 10%). Raise an error if this is the case, to prevent him from inputting a percentage.
    
4. Finally, the program should only show one number with a maximum of two digits after the decimal point.


```python
montant_initial=input("Combien d'argent souhaitez vous placer?")
years=input("pendant combien d'années souhaitez vous placer votre argent?")
interet=input("avez-vous le droit à un taux d'intérêt?")

montant_initial=int(montant_initial)
years=int(years)
interet=float(interet)

montant_final=montant_initial*(1+interet)**years

print("le montant total que vous aurez après avoir déposé {:.2f} € à la fin des {} années sera de {:.2f} €".format(montant_initial, years, montant_final))

```

```python
try:
    montant_initial=float(input("Combien d'argent souhaitez vous placer? (Veuillez saisir un nombre entier positif)"))
    years=float(input("pendant combien d'années souhaitez vous placer votre argent?"))
    interet=float(input("avez-vous le droit à un taux d'intérêt?"))

except ValueError:
    raise ValueError("Merci d'entrer uniquement des nombres!")

else:

    if (montant_initial <0) or (years<0) or (interet<0):
        print("veuillez saisir un nimbre entier")
    if (interet>1):
        print("Merci d'entrer une valeur d'intérêt inférieur à 1")

total=montant_initial*(1+interet)**years

print("le montant total que vous aurez après avoir déposé {:.2f} € à la fin des {} années sera de {:.2f} €".format(montant_initial, years, total))
    
```
