```python
#j'ajoute un argument
#name comme argument

def hi(name):
    greetings="Hello {}".format(name)
    return greetings
```

```python
print(hi("ANTOINE"))
print(hi("Julia"))
print(hi("Thomas"))
```

```python
def hi(name=""):
    if name=="":
        greetings="Hello Mr Nobody"
    else:
        greetings="Hello {}".format(name)
    return greetings
```

```python
print(hi("Julia"))
print(hi())
#je n'ai pas besoin de spécifier un argument pour que ma fonction fonctionne
```
