```python
def quiz(question, vrai_reponse, nb_de_chances):#je définie ma fonction quizz, dans laquelle j'ai 3 variables
    while True: #tant que je suis dans ma condition
        reponse_utilisateur = input(question) 
        if reponse_utilisateur != vrai_reponse:
            nb_de_chances -= 1
            print(f"Dommage ! Il te reste {nb_de_chances} chances")
            if nb_de_chances == 0:
                print("Oh non ! Tu as perdu le jeu...")
                break #j'ai plus de chances, j'arrête ma boucle
        else :
            print("Bravo tu as répondu correctement")
            break 
    return nb_de_chances #je dis le nb de chances restantes

#ceci est notre quizz
nb_de_chances = 3
print("Voici notre quiz, tu as trois chances !")

#il y a 3 questions
questions = [("Combien de fois la France a gagné la coupe du monde ?", "2"), 
             ("Quand a été fondé Apple ?", "1976"),
             ("Qui a fondé SpaceX ?", "elon musk")]

for question, answer in questions:
    nb_de_chances = quiz(question, answer, nb_de_chances)
    if nb_de_chances == 0:
        break

if nb_de_chances > 0 :
    print("Tu as gagné !!!")
```
