```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler, LabelEncoder
from sklearn.compose import ColumnTransformer

#sklearn = module qui permet le ML. On peut aussi faire l'import au fur et à mesure
```

```python
path_to_file="D:/Profils/NLefort/Desktop/JEDHA"

df=pd.read_csv(path_to_file+"/Data.csv")

df.head()
```

On veut prédire le comportement d'achat

# Etape 3 : Feagure/traget split


```python
#Je veux toutes les lignes et toutes les colonnes.
#soit je lis [:,0:4] --> en allant de gauche à droite. ou alors [:,:-1] pour dire que je lis de droite à gauche en incluant toutes les colonnes

"""Méthode 1
x=df.iloc[:,:-1]
#je veux colonne -1 en incluant le dataframe
y=df.iloc[:, -1:]
"""
"""Méthode 2"""
x=df.loc[:, ["Country", "Age", "Salary"]]
y=df.loc[:, ["Purchased"]]

y.head()
```

# Etape 4

```python
#train_test_split !
#test_size = entre 20% et 30% de mes données vont aller dans le test set
#random state = moyen pour fixer la méthode de choix aléatoires.
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=0)

x_train.head()
```

```python
# notanumber
# gestion des valeurs manquantes
# créer des instances de classes = je dis que ma stratégie est de remplacer par "médiane"
# une classe commence par une lettre en MAJ, soit SimpleImputer = classe
# une fonction est en min, comme imputer

imputer=SimpleImputer(strategy="median")
imputer
```

```python
#j'applique ma classe sur les données = fit.transform
# je transforme mes valeurs
imputer.fit_transform(x_train.iloc[:,1:])
```

```python
x_train.iloc[:,1:]=imputer.fit_transform(x_train.iloc[:,1:])
x_train.head()
#mes valeurs NaN sont remplacés par la médiane
```

```python
#Scaling & encoding
# créer un mini pipeline - mes données passent d'abord par sklearn puis encodeur
# je défini numeric fetaure = liste d'index de colonnes. j'ai des valeurs numériques dans colonnes 14 et 2

numeric_features=[1,2]
numeric_transformer= StandardScaler()

#pour mes variables catégorielles
categorical_features = [0]
categorical_transformer = OneHotEncoder ()

#le fait de mettre ma classe (par exemple, StandardScaler() sans rien entre les () = j'applique les paramètres par défaut à ma classe)

#création pipeline pré-processing =column transformer
#je transforme mes colonnes en 2 fonctionnalités. fonctionnalité catégorielle et fonctionnalité numérique.
#ces 2 transformers prennent la forme de tupples en fonction du nb de transformer que l'on aura.
feature_encoder = ColumnTransformer(
    transformers=[
        ('cat', categorical_transformer, categorical_features),
         #ce que prend le tupple = peut prendre n'importe quel nom, nommer le transformer auquel cela correspond, sur quelle colonne cela s'applique
        ('num', numeric_transformer, numeric_features)
    ]
)

#il faut que j'assigne sur x_train

x_train=feature_encoder.fit_transform(x_train)
x_train
```

```python
#j'ai bien une nouvelle matrice avec mes nouvelles valeurs
```
