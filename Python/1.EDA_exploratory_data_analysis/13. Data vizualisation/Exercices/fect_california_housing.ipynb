# Importation des librairies



```python
# Importation des bibliothèques
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import fetch_california_housing # fonction de sklearn qui charge le dataset California Housing.
```

```python
# Chargement du dataset California Housing
data = fetch_california_housing() # charge les données sous forme de dictionnaire.
# Création d'un DataFrame Pandas
cali_houses = pd.DataFrame(data.data, columns=data.feature_names) # crée un DataFrame avec les colonnes correctement nommées.
```

# Questions Bonus

### Quelle est la relation entre les variables, la plus forte  ?
- Grace à la mtrice de corrélation nous pouvons déterminer les fortes corrélations (proche de 1 ou -1).


```python
# Visualisation des corrélations entre les variables
plt.figure(figsize=(10, 8))
sns.heatmap(cali_houses.corr(), annot=True, cmap="coolwarm", fmt='.2f')
plt.title("Matrice de Corrélation des Variables")
plt.savefig("correlation_matrix.png")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Création du scatter plot
plt.figure(figsize=(8, 6))
sns.scatterplot(x=cali_houses["AveRooms"], y=cali_houses["AveBedrms"], alpha=0.5)

# Ajout des labels et titre
plt.xlabel("Nombre moyen de pièces")
plt.ylabel("Nombre moyen de chambres")
plt.title("Corrélation entre le nombre de pièces et de chambres")

# Ajout d'une régression linéaire
sns.regplot(x=cali_houses["AveRooms"], y=cali_houses["AveBedrms"], scatter=False, color="red")

# Sauvegarde du visuel
plt.savefig("correlation_avebedrms_averooms.png")

# Affichage du graphique
plt.show()

```

- Si la corrélation est forte et positive, cela signifie que les maisons avec plus de pièces ont aussi plus de chambres.
- Si la corrélation est faible, cela signifie que le nombre de chambres et de pièces ne sont pas directement liés (ex : il y a des grandes maisons avec peu de chambres).
- Le scatter plot permet de voir la tendance visuellement et la matrice de corrélation donne un aperçu global.
