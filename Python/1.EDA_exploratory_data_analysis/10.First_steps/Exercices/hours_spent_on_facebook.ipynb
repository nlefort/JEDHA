```python
import pandas as pd
```

```python
df=pd.read_excel("Hours Spent on Facebook.xlsx")
df.head()
```

```python
#Calculer mon intervalle de confiance
#cl=[mu-E, m+E]
#E=t*s/sqrt(n)
mu=df.mean()
#mais je ne veux que ma colonne 2. donc mu=df.iloc[toutes mes lignes, colonne 1]
mu=df.iloc[:,1].mean()
#après on veut t, où pour avoir 95% de mon échantillon t=1.96
t=1.96
s=df.iloc[:,1].std()
#taille de mon dataset = len()
sqrt_n=len(df)**0.5 

E=t*s/sqrt_n
print(E)
```

```python
cl=[mu-E,mu+E]
cl
```
